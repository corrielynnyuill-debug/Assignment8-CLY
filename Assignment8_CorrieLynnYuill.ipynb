```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Fish Market Dataset from GitHub
url = "https://raw.githubusercontent.com/Ankit152/Fish-Market/main/Fish.csv"
fish = pd.read_csv(url)

# Perform Initial EDA for dataset structure
print(fish.head())
print('-'*80)
print(fish.info())
print('-'*80)
print(fish.describe())
print('-'*80)

# Additional EDA Visualizations

# Histogram distribution of numerical features
numeric_cols = fish.select_dtypes(include=['float64', 'int64']).columns

plt.figure(figsize=(14,10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(3,3,i)
    sns.histplot(fish[col], kde=True, bins=20, color='steelblue')
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()

print('-'*80)

# Distibution of categorical species
plt.figure(figsize=(10,6))
sns.countplot(data=fish, x='Species', color='blue')
plt.title('Distribution of Fish Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

print('-'*80)

# Correlation heatmap of num erical features
plt.figure(figsize=(10,6))
corr = fish.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
